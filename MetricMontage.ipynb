### 3. Process

Will use system args if no path provided in the cell above

```python
from __future__ import annotations

from pathlib import Path
from typing import Iterable, List, Tuple
from tqdm import tqdm
import sys
import json

from scenedetect import open_video, SceneManager
from scenedetect.detectors import ContentDetector


def scene_durations_seconds(
    video_path: str | Path,
    *,
    threshold: float = 27.0,
    min_scene_len_frames: int = 15,
) -> List[float]:
    """
    Detect scenes using PySceneDetect ContentDetector and return scene durations (seconds).
    - threshold: higher = fewer cuts detected, lower = more cuts detected
    - min_scene_len_frames: minimum scene length in frames
    """
    video = open_video(str(video_path))

    scene_manager = SceneManager()
    scene_manager.add_detector(
        ContentDetector(threshold=threshold, min_scene_len=min_scene_len_frames)
    )

    scene_manager.detect_scenes(video=video)
    scenes: List[Tuple[object, object]] = scene_manager.get_scene_list()

    durations: List[float] = []
    for start_tc, end_tc in scenes:
        # FrameTimecode objects support get_seconds()
        durations.append(float(end_tc.get_seconds() - start_tc.get_seconds()))

    return durations


def print_results(video_paths: Iterable[str | Path]) -> None:
    for p in video_paths:
        path = Path(p)
        print(path.name)

        try:
            durations = scene_durations_seconds(path)
        except Exception as e:
            print(f"[error] {e}")
            print()
            continue

        if not durations:
            print("(no scenes detected)")
        else:
            for i, d in enumerate(durations, start=1):
                # One line per scene: "index<TAB>duration"
                print(f"{i}\t{d:.3f}")

        print()  # blank line between videos


def parse_paths_from_argv(argv: List[str]) -> List[str]:
    """
    Supports:
      - multiple args: python script.py a.mp4 b.mp4
      - single JSON array arg: python script.py '["a.mp4","b.mp4"]'
    """
    if len(argv) == 1:
        return []

    if len(argv) == 2 and argv[1].strip().startswith("["):
        return json.loads(argv[1])

    return argv[1:]


if __name__ == "__main__":
    if not paths:
        arg = sys.argv
        if arg:
            # Use system args if no path provided in the cell above
            paths = parse_paths_from_argv(sys.argv)
        else:
            # Example fallback if running it with no args.
            paths = [
                "video1.mp4"
                "video2.mp4",
            ]
        
    print_results(paths)
```
